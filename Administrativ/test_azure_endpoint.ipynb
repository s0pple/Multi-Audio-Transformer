# Import Required Libraries
Import necessary Python libraries such as urllib.request, json, os, and ssl for handling HTTP requests and JSON data.

```python
import urllib.request
import json
import os
import ssl
```

# Allow Self-Signed HTTPS
Define and call a function to bypass server certificate verification, which is important for some Azure environments.

```python
def allowSelfSignedHttps(allowed):
    # Bypass the server certificate verification on client side
    if allowed and not os.environ.get('PYTHONHTTPSVERIFY', '') and getattr(ssl, '_create_unverified_context', None):
        ssl._create_default_https_context = ssl._create_unverified_context

allowSelfSignedHttps(True)  # Important for some Azure environments
```

# Configure Endpoint URL and API Key
Set the Azure ML endpoint URL and API key for authentication.

```python
# --- YOUR LIVE CONFIGURATION ---
url = "https://multi-audio-transformer-zcalj.switzerlandnorth.inference.ml.azure.com/score"
api_key = "REDACTED_API_KEY"
# --------------------------------
```

# Prepare Test Data
Create a dictionary with test data including columns and sample rows for app_name, mouse_activity, hour, and day_of_week to simulate different scenarios.

```python
# Test data preparation
# We simulate 3 different situations to see if the model recognizes "Work" or "Relax".
data = {
    "input_data": {
        "columns": ["app_name", "mouse_activity", "hour", "day_of_week"],
        "data": [
            ["Excel", 8, 14, 2],      # Case 1: Excel, active, afternoon, Tuesday -> Expectation: Work
            ["Netflix", 1, 21, 6],    # Case 2: Netflix, little mouse, evening, Sunday -> Expectation: Relax
            ["VS Code", 9, 10, 1]     # Case 3: Coding, very active, morning, Monday -> Expectation: Work
        ]
    }
}
```

# Encode Data for Request
Encode the test data dictionary into a JSON string and then into bytes for the HTTP request body.

```python
# Encode data
body = str.encode(json.dumps(data))
```

# Create Request Headers
Set up HTTP headers including Content-Type and Authorization with the Bearer token.

```python
# Create headers with auth key
headers = {'Content-Type':'application/json', 'Authorization':('Bearer '+ api_key)}
```

# Send Request to Endpoint
Create and send the urllib request to the Azure endpoint, printing a status message.

```python
# Create request
req = urllib.request.Request(url, body, headers)

print(f"Sending request to: {url} ...")
```

# Process and Display Response
Handle the response, decode it, parse the JSON predictions, and print the results for each test case, including error handling for HTTP errors.

```python
try:
    # Send request
    response = urllib.request.urlopen(req)
    result = response.read()
    
    # Process response
    predictions = json.loads(result.decode("utf-8"))
    
    print("\nâœ… SUCCESS! The model has responded:\n")
    
    # Display results nicely
    for i, pred in enumerate(predictions):
        input_row = data['input_data']['data'][i]
        print(f"Input: {input_row}  -->  Classification: {pred}")

except urllib.error.HTTPError as error:
    print("\nâŒ ERROR: The request has failed.")
    print("Status Code: " + str(error.code))
    print(error.info())
    print(error.read().decode("utf8", 'ignore'))
```
